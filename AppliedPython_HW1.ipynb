# 1. Корутина для подсчета уникальных слов

````python
def unique_word_counter():
    unique_words = set()
    
    while True:
        text = yield

        if text is None:
            break
        
        words = text.split()
        unique_words.update(words)
    
    return len(unique_words)
````

````python
coro = unique_word_counter()
next(coro)
coro.send("hello world")
coro.send("hello again")
coro.send(None)
````

# 2. Корутина с ограничением по времени выполнения

````python
import time
import random
````

````python
def delay_yield():
    time.sleep(random.uniform(1, 3))
````

````python
def timed_collector(time_limit):
    start_time = time.time()
    total_sum = 0

    while True:
        value = yield
        delay_yield()
        total_sum += value

        if (time.time() - start_time > time_limit):
            break

    return total_sum
````

````python
coro = timed_collector(time_limit = 5)
next(coro)
coro.send(10)
coro.send(20)
coro.send(30)
coro.send(40)
coro.send(50)
````

# 3. Корутина для динамической фильтрации данных с прерыванием

````python
def dynamic_filter(mask):
    result = []

    while True:
        value = yield

        if value is None:
            break
        elif isinstance(value, (int, float)):
            if mask(value):
                result.append(value)
        else:
            mask = value
    
    return result
````

````python
coro = dynamic_filter(lambda x: x % 2 == 0)
next(coro)
coro.send(10)
coro.send(15)
coro.send(lambda x: x > 10)
coro.send(15)
coro.send(None)
````

# 4. Циклический генератор с обработкой исключений для отдельных заданий

````python
from collections import deque
````

````python
def cycle_task_queue(tasks):
    while tasks:
        task = tasks[0]
        
        try:
            value = next(task)
            yield value  
            tasks.rotate(-1)
            
        except (StopIteration, ValueError):
            tasks.popleft()
````

````python
def task_1():
    yield 1
    yield 2
    raise ValueError("Task 1 error")

def task_2():
    yield 3
    yield 4
    yield 5

tasks = deque([task_1(), task_2()])
coro = cycle_task_queue(tasks)
for item in coro:
    print(item)
````

````python
import csv
import json
````

````python
def merge_filtered_files(filter_func, output_filename, *generators):
    with open(output_filename, 'w') as file:
        coro = filter_and_write(filter_func, file)
        next(coro)
        
        for generator in generators:
            yield from (coro.send(line) for line in generator)
````

````python
def filter_and_write(filter_func, output_file):
    while True:
        line = yield
        
        if filter_func(line):
            output_file.write(line + '\n')
````

````python
def file_reader_txt(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()

def file_reader_csv(filename):
    with open(filename, 'r') as file:
        reader = csv.reader(file)
        for row in reader:
            yield ', '.join(row)

def file_reader_json(filename):
    with open(filename, 'r') as file:
        data = json.load(file)
        for item in data:
            yield json.dumps(item)

def contains_keyword(line):
    return "keyword" in line

merge = merge_filtered_files(
    contains_keyword,
    'merged_output.txt',
    file_reader_txt('file1.txt'),
    file_reader_csv('file2.csv'),
    file_reader_json('file3.json')
)

for _ in merge:
    pass
````

# 6. Рекурсивное объединение и фильтрация вложенных данных

````python
def recursive_extract(data):
    if isinstance(data, str):
        yield data

    elif isinstance(data, dict):
        for value in data.values():
            yield from recursive_extract(value)

    elif isinstance(data, list):
        for item in data:
            yield from recursive_extract(item)
````

````python
def contains_keyword(s):
    return "keyword" in s

def filtered_strings(data, filter_func):
    for string in recursive_extract(data):
        if filter_func(string):
            yield string

nested_data = {
    "messages": [
        "This is a message with keyword.",
        "Just a simple message.",
        {"comments": ["Another keyword here.", "Nothing here."]},
        ["List with keyword.", "Another list entry."]
    ],
    "notes": "This is a single note without the word.",
    "logs": {
        "entries": [
            "Log entry with keyword.",
            {"subentry": "A nested keyword here."}
        ]
    }
}

for line in filtered_strings(nested_data, contains_keyword):
    print(line)
````
