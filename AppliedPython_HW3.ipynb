# 1. Сравнение подходов для синхронизации потоков

````python
import threading
import time
````

````python
def access_resource_with_lock(n, lock):
    with lock:
        print(f"Поток {n} начал выполнение")
        time.sleep(1)
        print(f"Поток {n} завершил выполнение")
````

````python
def access_resource_with_semaphore(n, semaphore):
    with semaphore:
        print(f"Поток {n} начал выполнение")
        time.sleep(1)
        print(f"Поток {n} завершил выполнение")
````

````python
lock = threading.Lock()
threads = [threading.Thread(target = access_resource_with_lock, args = (i, lock)) for i in range(5)]

for t in threads:
    t.start()
for t in threads:
    t.join()
````

````python
semaphore = threading.Semaphore(2)
threads = [threading.Thread(target = access_resource_with_semaphore, args = (i, semaphore)) for i in range(5)]

for t in threads:
    t.start()
for t in threads:
    t.join()
````

# 2. Очередь задач с задержкой на потоках

````python
from queue import PriorityQueue
from queue import Empty
import threading
import time
````

````python
def producer(queue):
    for i in range(5, 0, -1):
        queue.put(i)
        print(f"Производитель добавил: {i}")
        time.sleep(0.5)
````

````python
def consumer(queue):
    while True:
        try:
            item = queue.get(timeout = 2)
            print(f"Потребитель взял: {item}")
            time.sleep(1)
            queue.task_done()
        except Empty:
            print("Потребитель не получил новую задачу за 2 секунды, он завершает выполнение")
            break
````

````python
queue = PriorityQueue()

producer_thread = threading.Thread(target = producer, args = (queue,))
consumer_thread = threading.Thread(target = consumer, args = (queue,))

producer_thread.start()
consumer_thread.start()

producer_thread.join()
consumer_thread.join()
````

# 3. Ускорение обработки данных с помощью ProcessPoolExecutor / multiprocessing.Pool / modin

````python
import pandas as pd
import numpy as np
import time
from concurrent.futures import ProcessPoolExecutor
from multiprocessing import Pool, cpu_count
import warnings
warnings.filterwarnings('ignore')

df = pd.DataFrame({
    'col1': np.arange(1, 1000001),
    'col2': np.arange(1000000, 0, -1)
})
df.to_csv('test.csv', index = False)

def process_chunk(chunk):
    chunk['result'] = chunk['col1'].values ** 2 + 2 * chunk['col2'].values
    return chunk

for chunk_size in [25000, 200000, 500000]:
    start = time.time()
    df = pd.read_csv('test.csv')
    chunks = [df.iloc[i:i+chunk_size] for i in range(0, len(df), chunk_size)]
    with ProcessPoolExecutor(max_workers = 4) as executor:
        results = list(executor.map(process_chunk, chunks))
    pd.concat(results).to_csv(f'result_ppe_{chunk_size}.csv', index = False)
    print(f'ProcessPoolExecutor ({chunk_size}): {time.time() - start:.3f}')

start = time.time()
df = pd.read_csv('test.csv')
chunks = [df.iloc[i:i+25000] for i in range(0, len(df), 25000)]
with Pool(processes = 4) as pool:
    results = pool.map(process_chunk, chunks)
pd.concat(results).to_csv('result_pool.csv', index = False)
print(f'multiprocessing.Pool: {time.time() - start:.3f}')

start = time.time()
import modin.pandas as mpd
df = mpd.read_csv('test.csv')
df['result'] = df['col1'] ** 2 + 2 * df['col2']
df._to_pandas().to_csv('result_modin.csv', index = False)
print(f'Modin: {time.time() - start:.3f}')
````
